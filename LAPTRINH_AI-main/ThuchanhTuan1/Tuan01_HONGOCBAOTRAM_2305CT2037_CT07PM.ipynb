# Bài 2 — NumPy: mảng, trục, broadcasting, vector hóa *(40 phút)*

**Mục tiêu.** Hiểu `ndarray` là gì trong bộ nhớ, `axis=0` khác `axis=1` ở chỗ nào, chuẩn hóa dữ liệu bằng broadcalling **không dùng vòng lặp**, và **tự đo** xem vector hóa nhanh hơn bao nhiêu lần.

**Các bước.**

1. **2.1** — Tạo mảng `(1000, 5)` ngẫu nhiên có seed cố định bằng `np.random.default_rng(42)`; in `shape`, `ndim`, `dtype`, `nbytes`.
2. **2.2** — Ô `assert` kiểm tra kích thước và kiểu.
3. **2.3** — Trục: trên mảng nhỏ `3×4`, so sánh `sum(axis=0)` với `sum(axis=1)` — cả giá trị lẫn `shape`.
4. **2.4** — Rút gọn theo trục trên `X`: trung bình, độ lệch chuẩn từng cột; min/max toàn mảng.
5. **2.5** — **Chuẩn hóa từng cột bằng broadcasting**, một dòng: `(X - X.mean(axis=0)) / X.std(axis=0)`.
6. **2.6** — Ô `assert`: trung bình mỗi cột ≈ 0, độ lệch chuẩn mỗi cột ≈ 1.
7. **2.7 – 2.9** — Viết bản vòng lặp Python, **đo bằng `time.perf_counter`**, so sánh với bản vector hóa, báo cáo số lần nhanh hơn.

**Kết quả mong đợi.**

| Ô | Kết quả |
|---|---|
| 2.1 | `shape (1000, 5)` · `ndim 2` · `dtype float64` · `nbytes 40000` (= 5000 × 8 byte) |
| 2.3 | `S.sum(axis=0) = [12 15 18 21]`, shape `(4,)` — một số cho mỗi **cột** <br> `S.sum(axis=1) = [ 6 22 38]`, shape `(3,)` — một số cho mỗi **dòng** |
| 2.4 | Trung bình từng cột của `X` gần 0 nhưng **không** bằng 0 (quanh `±0.05`); độ lệch chuẩn quanh `1.0`; min ≈ `-3.65`, max ≈ `3.45` |
| 2.5 | Sau chuẩn hóa: các giá trị trung bình đều in ra là số 0 (một số có dấu trừ — `-0.` là "số 0 âm", hoàn toàn vô hại); độ lệch chuẩn in ra `[1. 1. 1. 1. 1.]` |
| 2.6 | Sai số trung bình cỡ `1e-17` — **không** bằng 0 tuyệt đối, đó là số dấu phẩy động bình thường |
| 2.9 | Vector hóa nhanh hơn vòng lặp **vài chục đến vài trăm lần** (máy phòng lab thường ~30–150×). Con số chính xác khác nhau tùy máy — **không** có đáp án cố định. Điều bắt buộc: hai cách cho **cùng kết quả**. |

````python
# 2.1 — tạo dữ liệu có seed cố định
import numpy as np

rng = np.random.default_rng(42)   # seed cố định -> mọi người chạy ra cùng một mảng

# TODO: tạo mảng X kích thước (1000, 5) lấy từ phân phối chuẩn.
#       Gợi ý: rng.standard_normal((1000, 5))  — chú ý dấu ngoặc: truyền MỘT tuple
X = rng.standard_normal((1000, 5))

print("shape :", X.shape)
print("ndim  :", X.ndim)
print("dtype :", X.dtype)
print("nbytes:", X.nbytes, "byte  =", X.nbytes / 1024, "KB")
print("Một phần tử chiếm:", X.itemsize, "byte")
````

````python
# 2.2 — tự kiểm tra
assert X.shape == (1000, 5), f"shape sai: {X.shape}"
assert X.ndim == 2, "Phải là mảng 2 chiều"
assert X.dtype == np.float64, f"dtype sai: {X.dtype}"
assert X.nbytes == 1000 * 5 * 8, "nbytes = số phần tử x 8 byte (float64)"

print("2.1 OK — 1000 x 5 x 8 byte =", X.nbytes, "byte")
````

````python
# 2.3 — TRỤC: axis=0 và axis=1 khác nhau chỗ nào
S = np.arange(12).reshape(3, 4)
print(S)
print()

print("S.sum(axis=0) =", S.sum(axis=0), "| shape", S.sum(axis=0).shape,
      "<- gộp CÁC DÒNG lại, còn một số cho mỗi CỘT")
print("S.sum(axis=1) =", S.sum(axis=1), "| shape", S.sum(axis=1).shape,
      "<- gộp CÁC CỘT lại, còn một số cho mỗi DÒNG")

# Cách nhớ: axis=i nghĩa là "trục i BIẾN MẤT khỏi shape".
#           S.shape = (3, 4);  sum(axis=0) -> (4,);  sum(axis=1) -> (3,)

# TODO: đoán shape TRƯỚC khi chạy, rồi in S.mean(axis=0) và S.max(axis=1)
print("S.mean(axis=0) =", S.mean(axis=0),
      "| shape", S.mean(axis=0).shape)

print("S.max(axis=1) =", S.max(axis=1),
      "| shape", S.max(axis=1).shape)
````

````python
# 2.4 — rút gọn theo trục trên X
print("Trung bình từng cột (axis=0):", X.mean(axis=0).round(4))
print("Độ lệch chuẩn từng cột      :", X.std(axis=0).round(4))
print("shape của kết quả           :", X.mean(axis=0).shape, "<- 5 số, một số / cột")

# TODO: in min và max của TOÀN BỘ mảng (gọi .min() / .max() KHÔNG truyền axis),
#       làm tròn 4 chữ số thập phân.
print("Min toàn bộ mảng:", round(X.min(), 4))
print("Max toàn bộ mảng:", round(X.max(), 4))
````

````python
# 2.5 — CHUẨN HÓA từng cột bằng BROADCASTING
#
#   X            có shape (1000, 5)
#   X.mean(axis=0) có shape      (5,)
#
# NumPy tự "kéo giãn" mảng (5,) thành (1000, 5) khi trừ — đó là broadcasting.
# Không sao chép dữ liệu, không vòng lặp Python.

# TODO: viết MỘT dòng: lấy X trừ trung bình từng cột rồi chia cho độ lệch chuẩn từng cột
Xs = (X - X.mean(axis=0)) / X.std(axis=0)

print("Trung bình sau chuẩn hóa  :", Xs.mean(axis=0).round(12))
print("Độ lệch chuẩn sau chuẩn hóa:", Xs.std(axis=0).round(12))
print("shape giữ nguyên          :", Xs.shape)
````

````python
# 2.6 — tự kiểm tra kết quả chuẩn hóa
assert Xs.shape == X.shape, "Chuẩn hóa không được đổi shape"
assert np.abs(Xs.mean(axis=0)).max() < 1e-9, "Trung bình từng cột phải xấp xỉ 0"
assert np.abs(Xs.std(axis=0) - 1).max() < 1e-9, "Độ lệch chuẩn từng cột phải xấp xỉ 1"

print(f"2.5 OK — sai số trung bình lớn nhất: {np.abs(Xs.mean(axis=0)).max():.2e}")
print(f"         sai số độ lệch chuẩn lớn nhất: {np.abs(Xs.std(axis=0) - 1).max():.2e}")
print("Hai con số trên KHÁC 0 — số dấu phẩy động là thế, nên mới so với 1e-9 chứ không so == 0.")
````

````python
# 2.7 — hai cách làm cùng một việc
import time

def chuan_hoa_vong_lap(A):
    '''Chuẩn hóa từng cột bằng vòng lặp Python thuần — cách CHẬM.'''
    n_dong, n_cot = A.shape
    out = np.empty_like(A)
    for j in range(n_cot):
        cot = [A[i, j] for i in range(n_dong)]
        tb = sum(cot) / n_dong
        do_lech = (sum((v - tb) ** 2 for v in cot) / n_dong) ** 0.5
        for i in range(n_dong):
            out[i, j] = (A[i, j] - tb) / do_lech
    return out


def chuan_hoa_vector(A):
    '''Chuẩn hóa từng cột bằng broadcasting — cách NHANH.'''
    # TODO: một dòng return, giống hệt ô 2.5
    return (A - A.mean(axis=0)) / A.std(axis=0)


print("Đã định nghĩa 2 hàm.")
````

````python
# 2.8 — ĐO THỜI GIAN bằng time.perf_counter
#
# Dùng perf_counter chứ không dùng %timeit: perf_counter là Python thuần,
# chạy được cả trong notebook lẫn trong file .py chạy bằng dòng lệnh.

t0 = time.perf_counter()
X_loop = chuan_hoa_vong_lap(X)
t_loop = time.perf_counter() - t0

# Bản vector hóa chạy quá nhanh để đo chính xác trong 1 lần
# -> chạy N_LAP lần rồi lấy trung bình.
N_LAP = 100
t0 = time.perf_counter()
for _ in range(N_LAP):
    X_vec = chuan_hoa_vector(X)
t_vec = (time.perf_counter() - t0) / N_LAP

print(f"Vòng lặp Python  : {t_loop * 1000:9.3f} ms")
print(f"Vector hóa NumPy : {t_vec  * 1000:9.4f} ms  (trung bình {N_LAP} lần chạy)")
print(f"Nhanh hơn        : {t_loop / t_vec:9.1f} lần")
````

````python
# 2.9 — tự kiểm tra: nhanh hơn thì được, nhưng phải RA CÙNG KẾT QUẢ
assert np.allclose(X_loop, X_vec), "Hai cách phải cho cùng một kết quả"
assert np.allclose(X_vec, Xs), "...và phải khớp với Xs tính ở ô 2.5"

speedup = t_loop / t_vec
assert speedup > 5, f"Chỉ nhanh hơn {speedup:.1f} lần — kiểm tra lại chuan_hoa_vector"

print(f"2.7 OK — hai cách cho cùng kết quả, vector hóa nhanh hơn {speedup:.0f} lần.")
print("Nhanh mà sai thì vô dụng: luôn kiểm tra kết quả trước, đo tốc độ sau.")
````

#### Vì sao phải sửa `mssv` thành chuỗi?

`read_csv` đoán kiểu dữ liệu từ nội dung. Cột `mssv` toàn chữ số nên nó đoán `int64` — đúng về mặt ký tự, **sai về mặt ngữ nghĩa**. Ba hậu quả cụ thể:

1. **Mất số 0 ở đầu.** Nếu trường có mã dạng `0052341`, `int64` sẽ biến nó thành `52341`. Dữ liệu mất, không khôi phục được.
2. **Phép toán vô nghĩa vẫn chạy.** `df["mssv"].mean()` không báo lỗi. Nguy hiểm nằm ở chỗ đó — sai mà im lặng.
3. **Ghép bảng hỏng.** Tuần sau nối bảng này với bảng điểm mà một bên `int64`, một bên chuỗi, `merge` sẽ khớp 0 dòng và không ai hiểu vì sao.

**Quy tắc chung: mã số, số điện thoại, mã bưu chính, CCCD — luôn đọc dưới dạng chuỗi.**

````python
# 3.3b — SỬA: ép mssv thành chuỗi ngay khi đọc
# TODO: nạp lại df từ CSV_PATH, thêm tham số dtype={"mssv": str}
df = pd.read_csv(CSV_PATH, dtype={"mssv": str})

print("dtype sau khi sửa:", df["mssv"].dtype)
print("Một giá trị       :", repr(df["mssv"].iloc[0]))
print("shape             :", df.shape)
````

### Năm câu hỏi chọn lọc

**Quy tắc bắt buộc:** chỉ dùng `.loc`, `.iloc` và mặt nạ boolean. **Không** `for`, **không** `while`, **không** `iterrows()`, **không** `apply()` duyệt từng dòng.

Ba điều hay sai khi viết mặt nạ boolean:

| Sai | Đúng | Vì sao |
|---|---|---|
| `df[a == 1 and b == 2]` | `df[(a == 1) & (b == 2)]` | Với Series phải dùng `&` `\|` `~`, không dùng `and` `or` `not` |
| `df[a == 1 & b == 2]` | `df[(a == 1) & (b == 2)]` | `&` có độ ưu tiên cao hơn `==` → **bắt buộc** có ngoặc đơn quanh từng điều kiện |
| `df[df.gpa == None]` | `df[df["gpa"].isna()]` | `NaN` không bằng bất cứ thứ gì, kể cả chính nó |

````python
# CÂU 1 — Có bao nhiêu sinh viên khoa "Kỹ thuật Công nghệ" đạt GPA >= 3.2?

# TODO: tạo mặt nạ boolean gồm HAI điều kiện nối bằng &.
#       Nhớ bọc ngoặc đơn quanh từng điều kiện.
mask_1 = (
    (df["khoa"] == "Kỹ thuật Công nghệ")
    & (df["gpa"] >= 3.2)
)

cau_1 = df.loc[mask_1, ["mssv", "ho_ten", "nam_hoc", "gpa"]]

print("Số sinh viên:", len(cau_1))
print("GPA trung bình của nhóm này:", round(cau_1["gpa"].mean(), 3))
cau_1.sort_values("gpa", ascending=False).head(5)
````

````python
# CÂU 2 — Số giờ tự học trung bình mỗi tuần theo từng khoa?

# TODO: gom nhóm theo "khoa", lấy trung bình cột "gio_tu_hoc", sắp xếp giảm dần.
#       Gợi ý: df.groupby(...)[...].mean().sort_values(ascending=False)
cau_2 = (
    df.groupby("khoa")["gio_tu_hoc"]
      .mean()
      .sort_values(ascending=False)
)

print(cau_2.round(2))
print()
print("Cao nhất :", cau_2.idxmax(), "-", round(cau_2.max(), 2), "giờ/tuần")
print("Thấp nhất:", cau_2.idxmin(), "-", round(cau_2.min(), 2), "giờ/tuần")
print("Chênh lệch:", round(cau_2.max() - cau_2.min(), 2), "giờ/tuần")
````

````python
# CÂU 3 — Có bao nhiêu sinh viên NĂM 3 quê ở "TP. Hồ Chí Minh"?

# TODO: mặt nạ hai điều kiện: nam_hoc bằng 3 VÀ tinh_tp bằng "TP. Hồ Chí Minh"
mask_3 = (
    (df["nam_hoc"] == 3) & (df["tinh_tp"] == "TP. Hồ Chí Minh")
)


so_luong_3 = int(mask_3.sum())     # mặt nạ boolean: True = 1, nên .sum() là phép đếm
tong_nam_3 = int((df["nam_hoc"] == 3).sum())

print("Sinh viên năm 3 quê TP. Hồ Chí Minh:", so_luong_3)
print("Tổng sinh viên năm 3               :", tong_nam_3)
print("Chiếm:", round(100 * so_luong_3 / tong_nam_3, 1), "%")
````

````python
# CÂU 4 — 10 sinh viên có GPA cao nhất là ai?

# TODO: lấy 10 dòng GPA cao nhất, hiển thị các cột mssv, ho_ten, khoa, nam_hoc, gpa.
#       Gợi ý: df.nlargest(10, "gpa")[[...]]
#       Hoặc:  df.sort_values("gpa", ascending=False).iloc[:10][[...]]
cau_4 = df.nlargest(10, "gpa")[["mssv", "ho_ten", "khoa", "nam_hoc", "gpa"]]

print("Số dòng:", len(cau_4))
print("GPA cao nhất:", cau_4["gpa"].max(), "| GPA thấp nhất trong top 10:", cau_4["gpa"].min())
cau_4
````
